```python
import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles
from pydantic import color
%matplotlib inline
```

```python
import numpy as np
```

```python
x = np.arange(0,10)
y = np.arange(11,21)
```

```python
a = np.arange(40,50)
b = np.arange(50,60)
```

```python
plt.scatter(x,y,c='r')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Scatter Plot")
plt.savefig("Scatter Plot.png")
```

```python
y = x*x
plt.plot(x,y,'g*',linestyle = '-.',linewidth = 2,markersize = 12)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("2D line plot")
```

```python
# SubPlots
plt.subplot(2,2,1)
plt.plot(x,y,'r--')
plt.subplot(2,2,2)
plt.plot(x,y,'g-.')
plt.subplot(2,2,3)
plt.plot(x,y,'b*-')
```

```python
x = np.arange(0,4*np.pi,0.1)
y = np.sin(x)
plt.plot(x,y)
plt.title("Sine Plot")
```

```python
x = np.arange(0,5*np.pi,0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y_sin,'r-.')

plt.subplot(2,1,2)
plt.plot(x,y_cos,'g--')

plt.show()
```

```python
x= [2,8,10]
y= [11,16,9]
x2 = [3,9,11]
y2 = [6,15,7]

plt.bar(x,y,align='center')
plt.bar(x2,y2,color='red')

plt.title('Bar Graph')
plt.ylabel('Y axis')
plt.xlabel('X axis')

plt.show()
```

```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a, bins=200)
plt.title('Histogram')
plt.show()
```

```python
data = [np.random.normal(0,std,100) for std in range(1,4)]
plt.boxplot(data,vert=True,patch_artist=True)
plt.show()
```

```python
labels = 'python','c++','java','ruby'
sizes = [215,130,245,210]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode = (0.1,0,0,0)

plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct='%1.1f%%',shadow=True)

plt.axis('equal')
plt.show()
```
